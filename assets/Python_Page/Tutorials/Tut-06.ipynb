## 1. File Handling

### Q1. Log File Analyzer

````python
with open("log.txt", "w") as f:
    f.write("""System started successfully
User logged in
ERROR: Invalid password
Connection established
ERROR: Database not found
User logged out
ERROR: Timeout occurred
System shutdown""")
````

````python
error_count = 0

with open("log.txt", "r") as f:
    with open("errors.txt", "w") as out:
        for line in f:
            if "ERROR" in line:
                error_count += 1
                out.write(line)

print("Total errors:", error_count)
````

### Q3. Append vs Write Demonstration

````python
# Step 1 — Create data.txt
with open("data.txt", "w") as f:
    f.write("Python\nJava\nC++")

# Step 2 — Append
with open("data.txt", "a") as f:
    f.write("\nJavaScript")

print("After append:")
with open("data.txt") as f:
    print(f.read())

# Step 3 — Overwrite
with open("data.txt", "w") as f:
    f.write("JavaScript")

print("\nAfter write:")
with open("data.txt") as f:
    print(f.read())
````

### Q4. stats.py Module

stats.py
```
def mean(numbers):
    return sum(numbers) / len(numbers)

def maximum(numbers):
    return max(numbers)

def minimum(numbers):
    return min(numbers)
```

````python
with open("stats.py", "w") as f:
    f.write("""
def mean(numbers):
    return sum(numbers) / len(numbers)

def maximum(numbers):
    return max(numbers)

def minimum(numbers):
    return min(numbers)
""")
````

````python
import stats

data = [12, 45, 23, 67, 34]

print("Mean:", stats.mean(data))
print("Max:", stats.maximum(data))
print("Min:", stats.minimum(data))
````

### Q5. Namespace Conflict

mathops.py
```
def add(a, b):
    return a + b
```

stringops.py
```
def add(a, b):
    return a + b

````python
with open("mathops.py", "w") as f:
    f.write("""
def add(a, b):
    return a + b
""")

with open("stringops.py", "w") as f:
    f.write("""
def add(a, b):
    return a + b
""")
````

````python
import mathops as m
import stringops as s

print(m.add(10, 5))          # 15
print(s.add("Hello ", "World"))   # Hello World
````

### Q9. Copy vs Slice

````python
arr = np.array([10, 20, 30, 40])

copy_arr = arr.copy()
slice_arr = arr[1:3]

arr[1] = 99

print(arr)        # [10 99 30 40]
print(copy_arr)   # unchanged
print(slice_arr)  # changed because slice is a view\
````

### Q10. Weekly Temperature Plot

````python
import matplotlib.pyplot as plt
import numpy as np

temps = np.array([22, 24, 19, 23, 25, 21, 20])
days = np.array([1,2,3,4,5,6,7])

plt.plot(days, temps)
plt.title("Weekly Temperature")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.show()
````

### Q13. Kinematics

````python
import matplotlib.pyplot as plt
import numpy as np

time = np.array([1,2,3,4])
speed = np.array([10,20,30,40])
distance = np.array([10,40,90,160])

plt.subplot(2,1,1)
plt.plot(time, speed)
plt.title("Speed vs Time")

plt.subplot(2,1,2)
plt.plot(time, distance)
plt.title("Distance vs Time")

plt.tight_layout()
plt.show()
````

## Bonus: Mixed File Analyzer

````python
with open("mixed.txt", "w") as f:
    f.write("""ERROR Disk full
User login
75
ERROR Network down
hello
100
ERROR Access denied
42""")
````

````python
with open("mixed.txt", "r") as f, \
     open("errors.txt", "w") as e, \
     open("numbers.txt", "w") as n:

    for line in f:
        line = line.strip()
        if "ERROR" in line:
            e.write(line + "\n")
        elif line.isdigit():
            n.write(line + "\n")
````
